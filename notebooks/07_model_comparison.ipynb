```python
import pandas as pd
import matplotlib.pyplot as plt

comparison_df = pd.DataFrame({
    "Model": [
        "Logistic Regression",
        "Decision Tree",
        "Random Forest"
    ],
    "Accuracy": [0.846, 0.846, 0.846],
    "Precision": [0.500, 0.500, 0.500],
    "Recall": [0.150, 0.350, 0.050],
    "F1 Score": [0.231, 0.412, 0.091]
})

comparison_df.set_index("Model").plot(
    kind="bar",
    figsize=(10, 6)
)

plt.title("Model Performance Comparison")
plt.xlabel("Machine Learning Model")
plt.ylabel("Score")
plt.ylim(0, 1)
plt.xticks(rotation=0)
plt.legend(title="Metric")
plt.tight_layout()
plt.show()
```

```python
import os

results_dir = "/content/drive/MyDrive/student-risk-ml-project/results"

os.makedirs(results_dir, exist_ok=True)

output_path = results_dir + "/model_comparison.csv"

comparison_df.to_csv(output_path, index=False)

print("Saved:", output_path)
```
