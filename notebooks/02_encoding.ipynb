```python
import pandas as pd

input_path = "/content/drive/MyDrive/student-risk-ml-project/data/processed/student_cleaned.csv"

df = pd.read_csv(input_path)

df.head()
```

```python
print("Shape:", df.shape)
print("Missing values:", df.isnull().sum().sum())
print("Duplicates:", df.duplicated().sum())
```

```python
categorical_cols = df.select_dtypes(include="object").columns.tolist()

print("Number of categorical columns:", len(categorical_cols))
print(categorical_cols)
```

```python
for col in categorical_cols:
    print(f"\n{col}: {df[col].unique()}")
```

```python
binary_cols = [
    "school",
    "sex",
    "address",
    "famsize",
    "Pstatus",
    "schoolsup",
    "famsup",
    "paid",
    "activities",
    "nursery",
    "higher",
    "internet",
    "romantic"
]

multi_cols = [
    "Mjob",
    "Fjob",
    "reason",
    "guardian"
]

print("Binary columns:", len(binary_cols))
print("Multi-category columns:", len(multi_cols))
```

```python
encoded_df = df.copy()
```

```python
binary_mappings = {
    "school": {"GP": 0, "MS": 1},
    "sex": {"F": 0, "M": 1},
    "address": {"R": 0, "U": 1},
    "famsize": {"LE3": 0, "GT3": 1},
    "Pstatus": {"A": 0, "T": 1},
    "schoolsup": {"no": 0, "yes": 1},
    "famsup": {"no": 0, "yes": 1},
    "paid": {"no": 0, "yes": 1},
    "activities": {"no": 0, "yes": 1},
    "nursery": {"no": 0, "yes": 1},
    "higher": {"no": 0, "yes": 1},
    "internet": {"no": 0, "yes": 1},
    "romantic": {"no": 0, "yes": 1}
}

for col, mapping in binary_mappings.items():
    encoded_df[col] = encoded_df[col].map(mapping)
```

```python
for col in binary_cols:
    print(col, encoded_df[col].unique())
```

```python
print(
    "Missing values after binary encoding:",
    encoded_df[binary_cols].isnull().sum().sum()
)
```

```python
encoded_df = pd.get_dummies(
    encoded_df,
    columns=multi_cols,
    drop_first=False,
    dtype=int
)
```

```python
print("Before encoding:", df.shape)
print("After encoding:", encoded_df.shape)
```

```python
remaining_categorical = encoded_df.select_dtypes(include="object").columns.tolist()

print("Remaining categorical columns:", remaining_categorical)
```

```python
print(
    "Missing values after encoding:",
    encoded_df.isnull().sum().sum()
)
```

```python
print(
    "Duplicate rows after encoding:",
    encoded_df.duplicated().sum()
)
```

```python
print(encoded_df.dtypes.value_counts())
```

```python
all_numeric = all(
    pd.api.types.is_numeric_dtype(encoded_df[col])
    for col in encoded_df.columns
)

print("All columns numeric:", all_numeric)
```

```python
import os

processed_dir = "/content/drive/MyDrive/student-risk-ml-project/data/processed"

os.makedirs(processed_dir, exist_ok=True)

output_path = processed_dir + "/student_encoded.csv"

encoded_df.to_csv(output_path, index=False)

print("Saved successfully:", output_path)
```

```python
check_df = pd.read_csv(output_path)

print("Saved shape:", check_df.shape)
print("Missing values:", check_df.isnull().sum().sum())
print(
    "Categorical columns:",
    check_df.select_dtypes(include="object").columns.tolist()
)
```

## **Member 2 — Categorical Encoding Summary**

The cleaned Student Performance dataset contained 17 categorical variables.

The categorical variables were separated into binary and multi-category nominal features.

Thirteen binary categorical variables were encoded using explicit 0/1 mappings.

Four multi-category variables — Mjob, Fjob, reason, and guardian — were transformed using one-hot encoding because these categories do not have a natural ordinal relationship.

After encoding:

- all features were numerical
- no missing values were introduced
- no duplicate records were introduced
- G1, G2, and G3 were retained for the feature-engineering stage
- no feature selection, scaling, or train/test splitting was performed

The resulting dataset was saved as `student_encoded.csv`.
